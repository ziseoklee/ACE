```python
import os
from datetime import datetime
import pandas as pd
import numpy as np


exp_type = "checker"  # Options: "checker", "gmm"
date = "20250903"
experiment_id = f"experiment_{exp_type}_{date}"

# Load results 
result_csv_path = f"{experiment_id}/results_{exp_type}.csv"
results_csv = pd.read_csv(result_csv_path)
print(f"Results loaded from {result_csv_path}")
stats_by_method = results_csv.groupby('Method').agg({
    'W1': ['mean', 'std'],
    'W2': ['mean', 'std'],
    'MMD': ['mean', 'std'],
    'TV': ['mean', 'std']
}).reset_index()
print(stats_by_method)
# Save stats to a new CSV
stats_csv_path = f"{experiment_id}/stats_summary_{exp_type}.csv"
stats_by_method.to_csv(stats_csv_path, index=False)
print(f"Stats summary saved to {stats_csv_path}")
```

```python
import numpy as np

def parse_scores(lines):
    # accept either a multiline string or an iterable of lines
    if isinstance(lines, str):
        lines = lines.splitlines()
    scores = []
    for line in lines:
        parts = line.strip().split()
        if not parts:
            continue
        try:
            score = float(parts[-1])  # rightmost number
            scores.append(score)
        except ValueError:
            # ignore lines that don't match the pattern
            continue
    return scores

lines = """len: 16: -8.35445
len: 18: -8.39448
len: 15: -7.81249
len: 14: -6.72482
len: 17: -8.85164
len: 16: -4.54711
len: 17: -6.92383
len: 19: -8.71386
len: 14: -7.03375
len: 17: -8.71458"""

scores = parse_scores(lines)

if not scores:
    print("No scores found.")
else:
    scores = np.array(scores)
    mean = scores.mean()
    std = scores.std()
    min_val = scores.min()

    print(f"Mean: {mean:.5f}")
    print(f"Std: {std:.5f}")
    print(f"Min: {min_val:.5f}")
```
